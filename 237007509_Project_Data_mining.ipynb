# (C) Dataset Selection

## Selected Dataset: WELFake Fake News Dataset (Text Data)

### Reasons for selection:
- Directly supports **text mining and classification**, which are core topics covered in the course (e.g., TF-IDF, bag-of-words, supervised learning).
- Enables application of **feature extraction techniques** such as n-grams and word embeddings for meaningful pattern discovery.
- Supports **advanced external techniques** not covered in class, including transformer-based models (BERT, RoBERTa) and topic modeling (LDA, BERTopic).
- Allows comparison between **traditional machine learning models** and **modern deep learning approaches** for fake news detection.
- The dataset size (approximately 72,000 articles) is large enough for realistic data mining experiments but still computationally manageable.
- The problem of fake news detection has strong **social relevance**, making the analysis meaningful and impactful.

### Trade-offs:
- The dataset consists of unstructured text data, which requires extensive preprocessing (tokenization, stop-word removal, vectorization).
- Transformer-based methods are computationally expensive and may require GPU resources or data sampling.
- Potential bias exists in how news articles are labeled as fake or real, which may affect model fairness and generalizability.
- There is limited opportunity for association rule mining or transactional pattern discovery compared to retail datasets.
- Results may be sensitive to domain-specific language and topics present in the dataset.





# D. Exploratory Data Analysis



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer

# Load dataset (update path if needed)
df = pd.read_csv("WELFake_Dataset.csv")

# Display basic info
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Check column names
df.columns
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Drop rows with missing text or label
df = df.dropna(subset=["text", "label"])
```

```python
# Class distribution
label_counts = df["label"].value_counts()

plt.figure()
label_counts.plot(kind="bar")
plt.title("Class Distribution (0 = Fake, 1 = Real)")
plt.xlabel("Label")
plt.ylabel("Count")
plt.show()
```

```python
# Text length (number of words per article)
df["text_length"] = df["text"].apply(lambda x: len(str(x).split()))

plt.figure()
plt.hist(df["text_length"], bins=50)
plt.title("Distribution of Article Length (in words)")
plt.xlabel("Number of words")
plt.ylabel("Frequency")
plt.show()
```

```python
# Compare text length by class
plt.figure()
sns.boxplot(x="label", y="text_length", data=df)
plt.title("Article Length by Class (0 = Fake, 1 = Real)")
plt.xlabel("Label")
plt.ylabel("Text Length")
plt.show()
```

```python
# Word frequency (Top 20 words)
vectorizer = CountVectorizer(stop_words="english", max_features=20)
X = vectorizer.fit_transform(df["text"])

word_freq = np.array(X.sum(axis=0)).flatten()
words = vectorizer.get_feature_names_out()

word_freq_df = pd.DataFrame({"word": words, "frequency": word_freq})
word_freq_df = word_freq_df.sort_values(by="frequency", ascending=False)

plt.figure(figsize=(8,5))
plt.bar(word_freq_df["word"], word_freq_df["frequency"])
plt.xticks(rotation=45)
plt.title("Top 20 Most Frequent Words")
plt.xlabel("Word")
plt.ylabel("Frequency")
plt.show()
```

```python
# Vocabulary size and sparsity
vectorizer_full = CountVectorizer(stop_words="english")
X_full = vectorizer_full.fit_transform(df["text"])

vocab_size = len(vectorizer_full.get_feature_names_out())
total_elements = X_full.shape[0] * X_full.shape[1]
non_zero = X_full.nnz
sparsity = 1 - (non_zero / total_elements)

print("Vocabulary size:", vocab_size)
print("Sparsity of document-term matrix:", sparsity)
```

```python
# Check duplicate articles
duplicate_count = df.duplicated(subset=["text"]).sum()
print("Number of duplicate articles:", duplicate_count)
```

# (E) Initial Insights and Direction

## Key Observations from EDA:
- The class distribution shows a mild imbalance between fake and real news articles, which may influence model performance.
- Article lengths vary widely, with fake news articles generally shorter than real news articles.
- The most frequent words are common news-related terms, while some distinctive terms appear more often in fake news articles.
- The vocabulary size is very large, resulting in a highly sparse document-term matrix when using bag-of-words or TF-IDF.
- Duplicate and near-duplicate articles exist, indicating potential redundancy in the dataset.
- Topics such as politics, health, and social issues appear frequently across both fake and real news, creating overlap between classes.

## Hypotheses:
- Traditional TF-IDF features may not capture the semantic differences between fake and real news effectively.
- Shorter articles with emotionally charged or sensational language are more likely to be labeled as fake news.
- Dimensionality reduction or dense embeddings (e.g., Word2Vec or BERT embeddings) will improve classification performance compared to sparse representations.
- Topic modeling can reveal latent themes that distinguish fake news from real news beyond surface-level word frequency.

## Potential Research Questions (RQs):
- How does classification performance change when using TF-IDF features versus transformer-based embeddings (BERT/RoBERTa)?
- Do topic modeling techniques (LDA or BERTopic) uncover distinct themes between fake and real news articles?
- How does article length influence the probability of being classified as fake news?
- What words or phrases contribute most strongly to model predictions according to explainable AI methods (e.g., SHAP or attention visualization)?
- Does removing duplicate or near-duplicate articles improve model generalization?

## Direction for External Techniques:
- Apply transformer-based models to capture contextual meaning in news articles.
- Use topic modeling to explore hidden structure and thematic differences in the dataset.
- Employ explainable AI methods to interpret classification decisions and assess potential bias.
- Compare results from traditional machine learning models with modern deep learning approaches to evaluate trade-offs between accuracy and interpretability.
